## 2. Question 1: How many survived overall?

```python
print(titanic['survived'].value_counts())sns.countplot(data=titanic, x='survived')plt.title('Survival Count (0=No, 1=Yes)')plt.show()
```

**Insight:** More passengers died than survived overall — roughly 62% did not survive in the full dataset.

## 3. Question 2: Did passenger class affect survival?

```python
sns.countplot(data=titanic, x='pclass', hue='survived')plt.title('Survival by Passenger Class')plt.show()
```

**Insight:** 1st class passengers had noticeably better survival odds than 3rd class — consistent with '1st class had better access to lifeboats' historically.

## 4. Question 3: What did the age distribution look like?

```python
sns.histplot(data=titanic, x='age', kde=True)plt.title('Age Distribution')plt.show()
```

**Insight:** Most passengers were young adults (20s-30s), with a smaller number of children and elderly passengers.

## 5. Question 4: Did gender affect survival?

```python
sns.countplot(data=titanic, x='sex', hue='survived')plt.title('Survival by Gender')plt.show()
```

**Insight:** Females had a much higher survival rate than males — consistent with the 'women and children first' evacuation policy.

## 6. Question 5: How do the numeric columns relate to each other?

```python
numeric_df = titanic[['survived','pclass','age','sibsp','parch','fare']]sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')plt.title('Correlation Heatmap')plt.show()
```

**Insight:** `pclass` and `fare` are negatively correlated (makes sense — lower class number costs more), and `survived` correlates negatively with `pclass` (better class = better survival odds) and positively with `fare` (higher fare = better survival odds).
