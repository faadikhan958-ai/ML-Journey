```python
import pandas as pd
data={
    'name':['ali','akbar','aslam'],
    'class':['A','B','C'],
    'marks':[30,40,50],
    'age':[23,45,66]
}
df=pd.DataFrame(data)
print(df)

```

```python
print(df[df['marks']>35])
```

```python
print(df[df['marks']>=30])
```

```python
result=df[(df['marks']>20) & (df['age']>40)]
print (result)
```

```python
result=df[(df['marks']>30)|(df['marks']>27)]
print(result)
```

```python

result=df[df['class'].isin(['A','B'])]
print (result)
```

```python
result=df[df['class'].isin(['A','C'])]
print(result)
```
