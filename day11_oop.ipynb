```python
class Student:
    def __init__(self,name,marks,grade):
        self.name=name
        self.marks=marks
        self.grade=grade
ali=Student("ali",87,'A')
sara=Student("sara",70,'B')
Ahmad=Student("ahmad",45,'C')
print(ali.name)
print(sara.marks)
```

```python
print(ali.name ,ali.marks,ali.grade)
print(sara.name ,sara.marks,sara.grade)
print(Ahmad.name ,Ahmad.marks,Ahmad.grade)
```

```python
class car:
    def __init__(self,model,color,price):
        self.model=model
        self.color=color
        self.price=price

car1=car('2020',"blue",12000000)
car2=car('2021',"white",1300000)
car3=car('2024',"black",1400000)

print(f"model:{car1.model},color:{car1.color},price:{car1.price}")
print(f"model:{car2.model},color:{car2.color},price:{car2.price}")
print(f"model:{car3.model},color:{car3.color},price:{car3.price}")
```

```python
class Bankaccount:
    def __init__(self,holder_name,balance):
        self.holder_name=holder_name
        self.balance=balance
Akbar=Bankaccount("Akbar",2560000)
Aslam=Bankaccount("Aslam",2445000)
Alizy=Bankaccount("alizy",240)
Ahbar=Bankaccount("Ahbar",240000)
Abeera=Bankaccount("Abeera",2304000)
Asad=Bankaccount("Asad",20000)

print(f"Holder_name:{Akbar.holder_name},Balance:{Akbar.balance}")
print(f"Holder_name:{Aslam.holder_name},Balance:{Aslam.balance}")
print(f"Holder_name:{Alizy.holder_name},Balance:{Alizy.balance}")
print(f"Holder_name:{Ahbar.holder_name},Balance:{Ahbar.balance}")
print(f"Holder_name:{Abeera.holder_name},Balance:{Abeera.balance}")
print(f"Holder_name:{Asad.holder_name},Balance:{Asad.balance}")
```

```python
class Movie:
    def __init__(self, title, genre, rating):
        self.title = title
        self.genre = genre
        self.rating = rating

movie1 = Movie("Inception", "Sci-Fi", 8.8)
movie2 = Movie("The Dark Knight", "Action", 9.0)

print(f"Title:{movie1.title}, Genre:{movie1.genre}, Rating:{movie1.rating}")
print(f"Title:{movie2.title}, Genre:{movie2.genre}, Rating:{movie2.rating}")
```
