```python
import numpy as np
import pandas as pd
```

```python
data={
    'class':['A','B','A','C','B','C'],
    'marks':[60,55,57,78,78,89]
}
df=pd.DataFrame(data)
result=df.groupby('class')['marks'].mean()
print(result)
```

```python
df.groupby('class')['marks'].mean()    # Har class ka average marks
df.groupby('class')['marks'].sum()     # Har class ka total marks
df.groupby('class')['marks'].count()   # Har class mein kitne students hain
df.groupby('class')['marks'].max()     # Har class ka highest marks
df.groupby('class')['marks'].min()     # Har class ka lowest marks
```
