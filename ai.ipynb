```python

import os
import sys
import time
import math
import random
import platform
from datetime import datetime

# ==========================================
#       FAWAD KHAN - PERSONAL AI SYSTEM
# ==========================================

GREEN = "\033[92m"
CYAN = "\033[96m"
BLUE = "\033[94m"
YELLOW = "\033[93m"
RED = "\033[91m"
PURPLE = "\033[95m"
WHITE = "\033[97m"
BOLD = "\033[1m"
RESET = "\033[0m"

def clear():
    os.system("cls" if os.name == "nt" else "clear")

def slow_print(text, delay=0.025):
    for char in text:
        print(char, end="", flush=True)
        time.sleep(delay)
    print()

def line():
    print(CYAN + "═" * 58 + RESET)

def banner():
    clear()
    print(GREEN + BOLD + r"""
╔══════════════════════════════════════════════════╗
║                                                  ║
║             F A W A D   K H A N                  ║
║                                                  ║
║          ⚡ PERSONAL AI SYSTEM ⚡                 ║
║                                                  ║
║       AI DEVELOPER  |  PYTHON ENGINEER           ║
║                                                  ║
╚══════════════════════════════════════════════════╝
""" + RESET)

    print(YELLOW + "       WELCOME TO THE FUTURE, FAWAD!" + RESET)
    print(CYAN + "       FAWAD AI — THE NEXT GENERATION" + RESET)
    line()
    print(WHITE + " SYSTEM   : FAWAD AI CORE" + RESET)
    print(WHITE + " DEVELOPER: FAWAD KHAN" + RESET)
    print(WHITE + " DATE     : " + datetime.now().strftime("%d-%m-%Y %H:%M:%S") + RESET)
    print(WHITE + " PLATFORM : " + platform.system() + RESET)
    line()

def loading(label, steps=30):
    print(PURPLE + "\n" + label + RESET)
    for i in range(steps + 1):
        percent = int(i / steps * 100)
        bar = "█" * (i // 2) + "░" * (steps // 2 - i // 2)
        print(
            f"\r{GREEN}[{bar}] {percent:3d}%{RESET}",
            end="",
            flush=True
        )
        time.sleep(0.035)
    print("\n")

def generate_data(n=100):
    random.seed(42)
    x = [random.uniform(1, 100) for _ in range(n)]
    y = [3.5 * v + 12 + random.uniform(-15, 15) for v in x]
    return x, y

def train_model(x, y, epochs=250, lr=0.0001):
    w = 0.0
    b = 0.0
    n = len(x)
    losses = []

    for epoch in range(epochs):
        dw = 0.0
        db = 0.0
        loss = 0.0

        for xi, yi in zip(x, y):
            pred = w * xi + b
            error = pred - yi
            loss += error ** 2
            dw += 2 * error * xi / n
            db += 2 * error / n

        w -= lr * dw
        b -= lr * db
        losses.append(loss / n)

    return w, b, losses

def show_training(losses):
    print(BLUE + BOLD + "\nTRAINING NEURAL ENGINE..." + RESET)
    for i in range(0, len(losses), 10):
        value = losses[i]
        bars = min(35, int(value / max(losses[0], 1) * 35))
        print(
            f"{GREEN}Epoch {i+1:03d} "
            f"| Loss: {value:10.3f} "
            f"| {'█' * bars}{RESET}"
        )
        time.sleep(0.025)

def predict(w, b, value):
    return w * value + b

def main():
    banner()

    slow_print(GREEN + "\n[BOOT] Initializing Fawad AI..." + RESET)
    loading("Loading AI modules")

    slow_print(CYAN + "[OK] Python environment detected." + RESET)
    slow_print(CYAN + "[OK] Developer: Fawad Khan" + RESET)
    slow_print(CYAN + "[OK] Data engine connected." + RESET)
    loading("Preparing machine learning model")

    x, y = generate_data()
    w, b, losses = train_model(x, y)

    show_training(losses)

    line()
    print(YELLOW + BOLD + "\n       FAWAD AI — MODEL REPORT" + RESET)
    line()
    print(f"{WHITE}Developer       : FAWAD KHAN{RESET}")
    print(f"{WHITE}Model           : Linear Regression{RESET}")
    print(f"{WHITE}Training samples: {len(x)}{RESET}")
    print(f"{WHITE}Weight          : {w:.4f}{RESET}")
    print(f"{WHITE}Bias            : {b:.4f}{RESET}")
    print(f"{WHITE}Final loss      : {losses[-1]:.4f}{RESET}")
    line()

    print(PURPLE + BOLD + "\n       LIVE PREDICTION CONSOLE" + RESET)
    print(CYAN + "Enter a number to predict. Type 'exit' to stop." + RESET)

    while True:
        user_input = input(GREEN + "\nFAWAD@AI >>> " + RESET).strip()

        if user_input.lower() in ("exit", "quit"):
            break

        try:
            value = float(user_input)
            result = predict(w, b, value)
            print(YELLOW + f"\n[FAWAD AI] Input: {value:.2f}" + RESET)
            print(GREEN + BOLD + f"[PREDICTION] {result:.2f}" + RESET)
        except ValueError:
            print(RED + "Please enter a valid number." + RESET)

    line()
    slow_print(GREEN + "FAWAD AI SHUTTING DOWN..." + RESET)
    print(CYAN + BOLD + "THANK YOU, FAWAD KHAN! SEE YOU NEXT TIME." + RESET)
    line()

if __name__ == "__main__":
    main()
```
