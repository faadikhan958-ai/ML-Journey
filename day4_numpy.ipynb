```python
students={
    'name':['ali','sara','ahmad'],
    'maths':[30,33,45],
    'english':[23,45,43],
    'urdu':[25,40,41]
}
df=pd.DataFrame(students)
print(df)
```

```python
df['total']=df['maths']+df['english']+df['urdu']
print(df)
```

```python
df['percentage']=df['total']/300*100
print(df)
```

```python
df=df.drop('percentage',axis=1)
print(df)
```
