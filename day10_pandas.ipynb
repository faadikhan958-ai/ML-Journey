```python
import pandas as pd
students=pd.DataFrame({
        'student_id':[1,2,3],
        'name':['ali','sara','ahmad']
})
marks=pd.DataFrame({
        'student_id':[1,2,3],
        'marks':[50,60,70]   
 })
merge=pd.merge(students,marks,on='student_id')
print(merge)
```

```python
import pandas as pd
data1={
'id':[1,2,3,4],
'name':['ali','akbar','ahsan','zara']
}
df1=pd.DataFrame(data1)
print(df1)
data2={
'id':[1,2,3],
'grade':['A','B','C']
}
df2=pd.DataFrame(data2)
print(df2)

merge=pd.merge(df1,df2,on='id',how='left')
print (merge)
```

```python
import pandas as pd
student=pd.DataFrame({
    'id':[1,2,3,4],
    'name':['ali','sara','saba','akram']
    })
grade=pd.DataFrame({
    'id':[1,2,3],
    'grade':['A','B','C']
})
merged_inner=pd.merge(student,grade,on='id',how='inner')
print(merged_inner)
merged_left=pd.merge(student,grade,on='id',how='left')
print(merged_left)
merged_outer=pd.merge(student,grade,on='id',how='outer')
print(merged_outer)
merged_right=pd.merge(student,grade,on='id',how='right')
print(merged_right)
```

```python
import pandas as pd
import numpy as np

students = pd.DataFrame({
    'id': [1, 2, 3, 4, 5],
    'name': ['Ali', 'Sara', 'Ahmad', 'Zara', 'Bilal'],
    'city': ['Lahore', 'Karachi', 'Lahore', 'Karachi', 'Lahore']
})

print(students)
```

```python
scores = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'marks': [85, np.nan, 70, 90]
})

print(scores)
```

```python
merge=pd.merge(students,scores,on='id',how='left')
print(merge)
```

```python
avg=merge['marks'].mean()
print(avg)
```

```python
merge['marks']=merge['marks'].fillna(avg)
print(merge)
```

```python
merge.groupby('city')['marks'].mean()
```
