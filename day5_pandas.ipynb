```python
import pandas as pd
data={
    'name':['ali','ahamad','akbar'],
    'marks':[90,30,70]
}
df=pd.DataFrame(data)
print(df)
def get_grade(x):
    if x>=70:
        return 'a'
    else:
        return 'b'
df['grade']=df['marks'].apply(get_grade)
print(df)
```

```python
data={
    'name':['ali','nazir'],
    'marks':[90,20]}
df=pd.DataFrame(data)
df['grade']=df['marks'].apply(lambda x: 'a' if x>=45 else 'b')
print(df)
```
