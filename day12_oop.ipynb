```python
#with super
class Hostelstudent(Student):
    def __init__(self, name, age ,room_no):
        super().__init__(name,age)
        self.room_no=room_no
h1=Hostelstudent("ali",20,120)
print(h1.name)        
print(h1.age)        
print(h1.room_no)        
```

```python
class Animal:
    def __init__(self,name,sound):
        self.name=name
        self.sound=sound
    def make_sound(self):
        print(f"{self.name} makes {self.sound}")
class Dog(Animal):
    def __init__(self, name, sound):
        super().__init__(name, sound)
       
    def make_sound(self):
        super().make_sound()
        print(f"{self.name} bark: {self.sound}")
A1=Dog("dog","Baaaow")
print(A1.name)
print(A1.sound)
A1.make_sound()
```

```python
class Vehicle:
    def __init__(self,brand,price):
        self.brand=brand 
        self.price=price
class Car(Vehicle):
    def __init__(self, brand, price,fuel_type):
        super().__init__(brand, price)
        self.fuel_type=fuel_type
v1=Car("toyata",1250000,"petrol")
print(v1.brand)
print(v1.price)
print(v1.fuel_type)
```

```python
class Employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
class Manager(Employee):
    def __init__(self, name, salary,team_size):
        super().__init__(name, salary)
        self.team_size=team_size
class Developer(Employee):
    def __init__(self, name, salary,programming_language):
        super().__init__(name, salary)
        self.programming_language=programming_language
m1=Manager("Ahmad",150000,9)
d1=Developer("Fawad",120000,"python")
print(m1.name , m1.salary , m1.team_size)
print(d1.name , d1.salary , d1.programming_language)
```
