```python
import pandas as pd 
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_val_score
rng=np.random.default_rng(42)
n=1000
churned=rng.integers(0,2,n)
df = pd.DataFrame({
    'tenure_months': rng.integers(1, 72, n),
    'monthly_charges': rng.normal(60, 20, n).round(1),
    'support_calls': rng.integers(0, 10, n),
    'exit_survey_done': churned,   
    'churned': churned,
})
print(df.head())
```

```python
x=df.drop(columns=df[['churned','exit_survey_done']])
y=df['churned']
model=RandomForestRegressor(n_estimators=100,random_state=42)
print("cross_val_score:",cross_val_score(model,x,y,cv=5).mean().round(3))
model.fit(x,y)
print(pd.Series(model.feature_importances_,index=x.columns).sort_values(ascending=False).round(3))


```
