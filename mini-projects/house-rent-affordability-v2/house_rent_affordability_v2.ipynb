```python
import pandas as pd
data={
    'salary': [15000, 17000, 19000, 45000, 12000, 8000, 35000],
    'rent': [1000, 5000, 6000, 12000, 10433, 20000, 9000]
}
df=pd.DataFrame(data)
df['rent_to_salary']=df['rent']/df['salary']
print(df['rent_to_salary'])
df['affordable']=(df['rent_to_salary']<=0.30).astype(int)
for i in range(len(df)):
    if df['affordable'][i] == 1:
        print(f"Row {i}: Affordable")
    else:
        print(f"Row {i}: Not Affordable")
```

```python
import pandas as pd

data = {
    'salary': [15000, 17000, 19000, 45000, 12000, 8000, 35000],
    'rent': [4000, 5000, 6000, 12000, 10433, 20000, 9000]
}
df = pd.DataFrame(data)

df['rent_to_salary'] = df['rent'] / df['salary']
df['affordable'] = (df['rent_to_salary'] <= 0.30).astype(int)

for i in range(len(df)):
    if df['affordable'][i] == 1:
        print(f"Row {i}: Affordable")
    else:
        print(f"Row {i}: Not Affordable")
```
