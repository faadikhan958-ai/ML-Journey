```python
import seaborn as sns
from sklearn.ensemble import RandomForestRegressor,HistGradientBoostingRegressor
from sklearn.model_selection import cross_val_score,GridSearchCV
mpg=sns.load_dataset('mpg')
print(mpg.head())
```

```python
x=mpg[['cylinders','displacement','horsepower','weight','acceleration','model_year']]
y=mpg['mpg']
default_model=HistGradientBoostingRegressor(random_state=42)
default_cv=cross_val_score(default_model,x,y,cv=5,scoring='neg_mean_absolute_error')
print('default_cv',round(- default_cv.mean(),2))
grid={
    'max_iter':[50,100,300],
    'learning_rate':[0.03,0.1,0.2]
     }
search=GridSearchCV(
    HistGradientBoostingRegressor(random_state=42),
    grid,
    cv=5,
    scoring='neg_mean_absolute_error'
)
search.fit(x,y)
print('best settings:',search.best_params_)
print('best cv mae:',round(-search.best_score_,2))


```
