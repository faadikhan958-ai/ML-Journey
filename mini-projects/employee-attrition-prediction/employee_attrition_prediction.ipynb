# Employee Attrition Risk Prediction

**Client scenario:** HR team wants to predict which employees might leave 
the company, based on years at company and satisfaction score.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
hr_data = {
    'years_at_company': [1, 2, 5, 6, 1, 3, 7, 2, 4, 1, 8, 2],
    'satisfaction_score': [3, 4, 8, 9, 2, 5, 9, 3, 7, 4, 8, 2],
    'left_company': [1, 1, 0, 0, 1, 0, 0, 1, 0, 1, 0, 1]
}
df=pd.DataFrame(hr_data)
x=df[["years_at_company","satisfaction_score"]]
y=df['left_company']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
model=LogisticRegression()
model.fit(x_train,y_train)
prediction=model.predict(x_test)
print(prediction)
print('actual:',y_test.values)
print('prediction:',prediction)
print("m:",model.coef_)
print('c:',model.intercept_)
print(df.corr())
accuracy=accuracy_score(y_test,prediction)
print('Accuracy:',accuracy)

      
```

```python
Employee=[[2,7]]
leave=model.predict(Employee)
print("will Leave?:",leave)
```
