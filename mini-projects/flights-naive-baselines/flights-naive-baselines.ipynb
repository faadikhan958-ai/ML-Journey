Ticket: “Ek airline agle mahine ke passengers ka andaza lagana chahti hai. Model banane se pehle batao ke sirf copy karna kitna kaam karta hai.”

```python
import seaborn as sns
flights=sns.load_dataset('flights')
passengers=flights['passengers']
print(passengers)
```

```python
last_month_errors=[]
last_year_errors=[]
for i in range(12,len(passengers)):
    actual=passengers[i]
    last_month_errors.append(abs(actual-passengers[i-1]))
    last_year_errors.append(abs(actual-passengers[i-12]))
print('copy month mae:',round(sum(last_month_errors)/len(last_month_errors)))
print('last_year_errors',round(sum(last_year_errors)/len(last_year_errors)))    
```
