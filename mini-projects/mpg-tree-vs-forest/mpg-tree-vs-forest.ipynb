```python
import seaborn as sns
import pandas as pd
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
mpg=sns.load_dataset('mpg')
print(mpg.isnull().sum())
```

```python
print(mpg.head())
```

```python
x= mpg[['cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model_year']]
y = mpg['mpg']
diff=abs(y-y.mean())
print("Baseline mae:",round(diff.mean(),2))
```

```python
tree=DecisionTreeRegressor(random_state=42)
tree.fit(x,y)
train_mae_tree=round(mean_absolute_error(y,tree.predict(x)),2)
print("train_mae_tree:",train_mae_tree)
tree_cv=cross_val_score(tree,x,y,cv=5,scoring='neg_mean_absolute_error')
print('tree cv:',round(-tree_cv.mean(),2))
forest=RandomForestRegressor(n_estimators=100,random_state=42)
forest.fit(x,y)
train_mae_forest=print('train_mae_forest:',round(mean_absolute_error(y,forest.predict(x)),2))
forest_cv=-cross_val_score(forest,x,y,cv=5,scoring='neg_mean_absolute_error')
print('forest_cv:',round(forest_cv.mean(),2))
```
