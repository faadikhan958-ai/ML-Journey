```python
import pandas as pd
import seaborn as sns
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_val_predict

mpg = sns.load_dataset('mpg')
mpg = mpg.dropna()
mpg = mpg.reset_index(drop=True)

x = mpg[['cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model_year']]
y = mpg['mpg']

forest = RandomForestRegressor(n_estimators=100, random_state=42)
pred = cross_val_predict(forest, x, y, cv=5)

results = mpg[['name', 'origin', 'weight', 'mpg']].copy()
results['pred'] = pred
results['error'] = abs(results['mpg'] - results['pred'])


print(results.groupby('origin')['error'].mean().round(2))
```

```python
print(results.sort_values('error', ascending=False).head(10))
```
