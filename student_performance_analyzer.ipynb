```python
import numpy as np
import pandas as pd
```

```python
names=['Ali','sara','Ahmad']
print (names)
```

```python
marks=[86,98,60]
print(marks)
```

```python
data={
    'name':names,
    'marks':marks
}
df=pd.DataFrame(data)
print(df)
```

```python
math = [85, 90, 70]
science = [80, 88, 65]
english=[40,67,78]
data={
    'name':names,
    'math':math,
    'science':science,
    'english':english
}
df=pd.DataFrame(data)
print(df)
```

```python
df['total']=df['math']+df['science']+df['english']
print(df)
```

```python
df['percentage']=(df['total']/300)*100
print(df)
```

```python
def get_grade(p):
    if p>=80:
        return 'A'
    elif p>=60:
        return 'B'
    elif p>=40:
        return 'c'
    else:
     return 'fail'
df['grade']=df['percentage'].apply(get_grade)
print(df)
```

```python
df.to_csv('student_project.csv',index=False)
```
