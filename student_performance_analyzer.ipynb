```python
class Student:
    def __init__(self, name, math, science, english):
        self.name = name
        self.math = math
        self.science = science
        self.english = english
    
    def get_total(self):
        return self.math + self.science + self.english
    
    def get_percentage(self):
        return (self.get_total() / 300) * 100
    
    def get_grade(self):
        p = self.get_percentage()
        if p >= 80:
            return 'A'
        elif p >= 60:
            return 'B'
        elif p >= 40:
            return 'C'
        else:
            return 'Fail'


class HonorStudent(Student):
    def __init__(self, name, math, science, english, extra_credit):
        super().__init__(name, math, science, english)
        self.extra_credit = extra_credit
    
    def get_total(self):
        return super().get_total() + self.extra_credit


# Normal student
ali = Student("Ali", 85, 80, 40)
print(f"{ali.name} -> Total: {ali.get_total()}, Percentage: {ali.get_percentage():.2f}, Grade: {ali.get_grade()}")

# Honor student
sara = HonorStudent("sara", 90, 88, 67, 15)
print(f"{sara.name} -> Total: {sara.get_total()}, Percentage: {sara.get_percentage():.2f}, Grade: {sara.get_grade()}")
```

```python
data = {
    'hours_studied': [1, 2, 3, 4, 5, 6],
    'marks': [30, 40, 50, 65, 75, 85]
}
df = pd.DataFrame(data)

X = df[['hours_studied']]
Y = df['marks']

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

model = LinearRegression()
model.fit(X_train, Y_train)

predictions = model.predict(X_test)
print("Actual:", Y_test.values)
print("Predicted:", predictions)
```

```python
mae=mean_absolute_error(Y_test,predictions)
print("mae:",mae)
```
